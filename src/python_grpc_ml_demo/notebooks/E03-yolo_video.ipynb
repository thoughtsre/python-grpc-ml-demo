```python
import cv2
from PIL import Image as I

import torch

import sys
sys.path.append("..")

from server import torch_pred_to_json, to_protobuf_prediction
from lib.ml_predictions_pb2 import *
```

```python
# Model
model = torch.hub.load('ultralytics/yolov5', 'yolov5s', pretrained=True)
```

```python
vid = cv2.VideoCapture("../../../assets/KFC.mp4")
```

```python
vid.isOpened()
```

```python
imgs = []

show_every = 24

cur_frame = 0

while True:
    
    ret, f = vid.read()
    
    cur_frame += 1
    
    if not ret:
        break
    
    img = I.fromarray(cv2.cvtColor(f, cv2.COLOR_BGR2RGB))
    
    
    if (cur_frame % show_every) == 0:
        print(f"Frame {cur_frame}")
        imgs.append(torch_pred_to_json(model([img])))
```
