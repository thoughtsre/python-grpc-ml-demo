```python
from torchvision.models import resnet18, ResNet18_Weights
from PIL import Image
```

```python
wts = ResNet18_Weights.IMAGENET1K_V1
model = resnet18(weights=wts)
model.eval()
preprocess = wts.transforms()
```

```python
img = Image.open("../assets/cat.jpg")
```

```python
img
```

```python
img_transformed = preprocess(img)
```

```python
img_transformed.expand(1, -1, -1, -1).shape
```

```python
pred = model(img_transformed.unsqueeze(0))
```

```python
pred.shape
```

```python
class_id = pred.argmax().item()
```

```python
wts.meta["categories"][class_id]
```
